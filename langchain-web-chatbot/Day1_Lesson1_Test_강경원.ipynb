#### 문항 7. 대화 맥락 유지

▶ 시나리오 <br/>
챗봇이 사용자의 이름을 기억하지 못한다는 피드백을 받았다. messages 리스트를 구성해 맥락을 유지하도록 수정해야 한다. <br/>
1) 사용자: "제 이름은 (본인 이름)입니다." <br/>
2) LLM: "안녕하세요, (본인 이름)님!" <br/>
3) 사용자: "제 이름이 뭐였죠?" ← 이름을 기억한 답변이 나와야 함 <br/>
▶ 요구사항 <br/>
1) 위 대화가 이어지도록 messages 리스트를 완성해 실행하시오. <br/>
2) 사용한 role 세 가지가 각각 어디에 쓰였는지 주석으로 표시하시오. <br/>


````python
def ask_llm(messages, temperature=0.7, max_tokens=50):
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages,
        temperature=temperature,
        max_tokens=max_tokens,
    )
    return response.choices[0].message.content

messages = [
    # 사용자
    {"role": "user", "content": "제 이름은 경원입니다."},
    # LLM
    {"role": "assistant", "content": "안녕하세요, 경원님!"},
    # 질문
    {"role": "user", "content": "제 이름이 뭐였죠?"}  
]

print(ask_llm(messages))
````

▶ 요구사항

2) 네 개의 문의를 모두 분류해 출력하시오.

````python
def classify_inquiry(text: str) -> str:
    messages = [
        {"role": "system", "content": "_결제/배송/환불/기타 중 하나로 분류"},
        {"role": "user", "content": text},
    ]
    return ask_llm(messages, temperature=0, max_tokens=10)
test_cases = [
    "어제 주문했는데 카드 결제가 두 번 되었어요.",
    "택배가 일주일째 안 와요.",
    "사이즈가 안 맞아서 반품하고 싶어요.",
    "영업시간이 어떻게 되나요?",
]

for text in test_cases:
    category = classify_inquiry(text)
    print(f"[{category:>4}] {text}")
````

#### 문항 9. temperature 비교 실험

▶ 시나리오 <br/>
temperature 값이 결과에 미치는 영향을 팀원에게 시연해야 한다. 정답이 정해진 질문으로는 차이가
드러나지 않으므로 창의성이 필요한 질문을 사용한다. <br/>
▶ 요구사항 <br/>
1) "AI 스타트업 이름 3개를 지어주세요"를 temperature=0과 temperature=1.2로 각각 두 번씩 호출
해 출력하시오. <br/>
2) 출력 아래 마크다운 셀에 관찰한 차이를 두 줄로 쓰시오. <br/>

````python
prompt = "AI 스타트업 이름 3개를 지어주세요"

for temp in [0.0, 1.2]:
    answer = ask_llm(
        [{"role": "user", "content": prompt}],
        max_tokens=300,
        temperature=temp,
    )
    print(f"temperature={temp}\n→ {answer}\n")
````

temperature = 0인 경우에는 직관적이고 창의적인 부분이 부족하며 항상 동일한 결과를 생성한다고 생각합니다. temperature=1.2로 작성한 경우에는 조금 더 창의성 있고 결과 생성을 여러번 하였을 때 다른 답을 생성한다고 생각합니다.

``` python
PRICE_INPUT = 0.15 # 입력 1M 토큰당 USD 
PRICE_OUTPUT = 0.60 # 출력 1M 토큰당 USD

```
▶ 요구사항 <br/>
1) 임의의 질문을 한 번 호출하시오. <br/>
2) 입력 토큰 수, 출력 토큰 수, 총 토큰 수를 출력하시오. <br/>
3) 위 단가를 적용해 원화 환산 비용을 출력하시오. <br/>

````python
PRICE_INPUT = 0.15 # 입력 1M 토큰당 USD 
PRICE_OUTPUT = 0.60 # 출력 1M 토큰당 USD 
USD_TO_KRW = 1400 # 환율

def cost_korean_won(usage):
    input_cost = usage.prompt_tokens / 1_000_000 * PRICE_INPUT
    output_cost = usage.completion_tokens / 1_000_000 * PRICE_OUTPUT
    return (input_cost + output_cost) * USD_TO_KRW

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": "LLM이란 무엇인가요?"}],
    temperature=0.7,
    max_tokens=200
)

u = response.usage
print(f"입력 {u.prompt_tokens} + 출력 {u.completion_tokens} = "
      f"총 {u.total_tokens} 토큰, 비용 약 {cost_korean_won(u):.4f}원")
````
