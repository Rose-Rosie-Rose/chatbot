```python
# RunnableLambda에 add_one을 넣음

from langchain_core.runnables import RunnableLambda

def add_one(x):
    return x + 1

r_add = RunnableLambda(add_one)
print(r_add.invoke(3))
print(r_add.batch([1, 2, 3]))
```

```python
from langchain_core.runnables import RunnableLambda

# 위에 r_add 함수와 동일한 코드 lambda 식으로 변경
r_add = RunnableLambda(lambda x: x + 1)
r_double = RunnableLambda(lambda x: x * 2)

chain_a = r_add | r_double
chain_b = r_double | r_add

# 연결 순서에 따라 결과가 다르게 나옴
print(chain_a.invoke(3))    # 8  3 + 1 = 4 * 2 = 8
print(chain_b.invoke(3))    # 7  3 * 2 + 1

# RunnableSequence 타입 -> 
print(type(chain_a))        # <class 'langchain_core.runnables.base.RunnableSequence'>
print(type(chain_b))
```

```python
from langchain_core.runnables import RunnableLambda, RunnableParallel, RunnablePassthrough

analyzer = RunnableParallel(
    original = RunnablePassthrough(),
    upper = RunnableLambda(lambda x : x.upper()),
    length = RunnableLambda(lambda x: len(x))
)

print(analyzer.invoke("hello"))
```

```python
from dotenv import load_dotenv
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI

# .env 파일의 API키 불러오기
load_dotenv()

prompt = ChatPromptTemplate.from_template("{topic}에 대해서 친절한 초등학교 선생님처럼 설명해줘")

model = ChatOpenAI(model = "gpt-4o-mini")
parser = StrOutputParser()

chain = prompt | model | parser

print(chain.invoke({"topic": "인공지능"}))
```

```python
results = chain.batch([
    {"topic": "블럭체인"},
    {"topic": "클라우드"}
])

# for r in results:
#     print("-", r)

for chunk in chain.stream({"topic": "양자컴퓨터"}):
    print(chunk, end="", flush=True)
```

```python
# 과제 1(기초, API 키 불필요)
# 문자열을 받아 양쪽 공백을 없애고, 대문자로 바꾼 뒤, 끝에 !를 붙이는 체인을 만드세요.
from langchain_core.runnables import RunnableLambda


strip_str = RunnableLambda(lambda x: x.strip())
upper_str = RunnableLambda(lambda x: x.upper())
append_str = RunnableLambda(lambda x: x + "!")

chain = strip_str | upper_str | append_str

print(chain.invoke("  hello world   "))
```

```python
# 과제 2 (중급, API 키 불필요)
# 숫자 리스트를 받아 합계, 평균, 최댓값을 한 번에 딕셔너리로 돌려주는 Runnable을 만드세요.
from langchain_core.runnables import RunnableLambda, RunnableParallel

num_change = RunnableParallel(
    total_num = RunnableLambda(lambda x: sum(x)),
    avg_num = RunnableLambda(lambda x: round(sum(x) / len(x))),
    max_num = RunnableLambda(lambda x: max(x))
)

print(num_change.invoke([4, 5, 6, 9]))
```

```python
# 과제 3 (심화, API 키 필요)
# 주제 하나를 넣으면 ① 한 줄 설명과 ② 관련 키워드 3개를 동시에 만들어 딕셔너리로 돌려주는 체인을 만드세요. 

from dotenv import load_dotenv
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import PydanticOutputParser
from langchain_openai import ChatOpenAI
from pydantic import BaseModel, Field

load_dotenv()

class MySchema(BaseModel):
    description: str = Field(description="주제에 대한 한 줄 설명")
    keywords: list[str] = Field(description="주제와 관련된 키워드 3개")


parser = PydanticOutputParser(pydantic_object=MySchema)

prompt = ChatPromptTemplate.from_messages([
    ("system", "주어진 주제를 한 줄로 설명하고 관련 키워드 3개를 생성하세요.\n{format_instructions}"),
    ("human", "{input}")
]).partial(format_instructions=parser.get_format_instructions())

model = ChatOpenAI(model="gpt-4o-mini")

chain = prompt | model | parser

result = chain.invoke({"input": "LangChain"})

print(result.description)
print(result.keywords)
```
